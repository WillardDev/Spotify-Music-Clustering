# Option 1: Music Clustering Project

**Goal:** Group songs by their audio features using unsupervised clustering so that hidden structure in the music catalogue can be surfaced for tasks such as recommendations and playlist generation.

**Steps covered in this notebook:**
- Load and explore the Spotify audio-feature dataset
- Preprocess the data (feature selection, duplicate removal, scaling)
- Reduce dimensionality with PCA for visualisation
- Apply **K-Means**, **Hierarchical clustering**, **DBSCAN**, and **Gaussian Mixture Models**
- Evaluate the clusters with the elbow method, silhouette scores, and dendrogram analysis
- Compare models and interpret the final clusters

**Data source:** `data/ML_spotify_data.csv` (10,000 tracks with Spotify audio features).

## 1. Setup and imports

```python
%matplotlib inline
import warnings
warnings.filterwarnings("ignore")

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans, AgglomerativeClustering, DBSCAN
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score
from sklearn.manifold import TSNE
from sklearn.neighbors import NearestNeighbors
from scipy.cluster.hierarchy import linkage, dendrogram

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (10, 5)
pd.set_option("display.float_format", lambda v: f"{v:.3f}")
```

## 2. Load and explore the data

```python
DATA_PATH = "data/ML_spotify_data.csv"
df = pd.read_csv(DATA_PATH)

print(f"Rows: {df.shape[0]:,}  |  Columns: {df.shape[1]}")
df.head()
```

```python
df.info()
print("-" * 40)
df.describe().T
```

```python
print(f"Missing values: {df.isna().sum().sum()}")
print(f"Duplicated rows: {df.duplicated().sum()}")
```

- 10,000 tracks with no missing values.
- One duplicate track exists and will be removed during preprocessing.
- Features span very different scales (e.g. loudness from about -60 dB to +1 dB, tempo from 0 to 224 BPM), so standardisation is required before clustering.

## 3. Preprocessing

### Feature selection

The project description lists nine audio features, but this particular dataset export (`data/ML_spotify_data.csv`) does **not** contain `acousticness` or `instrumentalness`. The available columns are `popularity`, `danceability`, `valence`, `energy`, `explicit`, `key`, `liveness`, `loudness`, `speechiness`, and `tempo` (plus `name` and `artists`). The two missing features cannot be derived reliably from the others, so clustering uses the numeric audio features that are actually present.

The nine features selected are:
- **`popularity`** - not strictly an audio feature, but included deliberately because a large part of the business use-case (recommendation, playlists, catalogue curation) is separating mainstream from niche material, and it drives some of the segments users care about.
- **`key`** - retained because it is one of the standard Spotify descriptors; it is close to nominal and mostly decorrelated, so in practice it contributes little to the distances and functions mainly as a sanity check on the clusters rather than a primary driver.
- The core audio dimensions: `danceability`, `energy`, `liveness`, `loudness`, `speechiness`, `tempo`, `valence`.

`explicit` is present in the data but deliberately **excluded** from the clustering: it is a binary flag rather than an audio-quality measure, and a binary variable can dominate Euclidean distances in a dataset this small. It is still useful afterwards as a descriptor to enrich cluster interpretation.

All selected features are standardised (zero mean, unit variance) so that no single feature dominates the distance calculations.

```python
AUDIO_FEATURES = [
    "popularity", "danceability", "valence", "energy", "key",
    "liveness", "loudness", "speechiness", "tempo",
]

data = df[AUDIO_FEATURES].drop_duplicates().reset_index(drop=True)
print(f"Rows after removing duplicates: {data.shape[0]:,}")

scaler = StandardScaler()
X = scaler.fit_transform(data)
X_scaled = pd.DataFrame(X, columns=AUDIO_FEATURES)
X_scaled.head()
```

## 4. Dimensionality reduction with PCA

PCA projects the 9-dimension feature space into uncorrelated components. It is used to inspect how much variance each component holds and to visualise the songs in 2D.

```python
pca = PCA().fit(X)
evr = pca.explained_variance_ratio_

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].plot(range(1, len(evr) + 1), np.cumsum(evr), "o-")
axes[0].axhline(0.8, ls="--", color="gray")
axes[0].set(title="Cumulative explained variance", xlabel="Components", ylabel="Cumulative variance")
axes[0].set_xticks(range(1, len(evr) + 1))
axes[1].bar(range(1, len(evr) + 1), evr)
axes[1].set(title="Variance per component", xlabel="Components", ylabel="Explained variance")
plt.tight_layout()
plt.show()

loadings = pd.DataFrame(pca.components_.T, index=AUDIO_FEATURES,
                        columns=[f"PC{i+1}" for i in range(len(evr))])
print("PC1 and PC2 loadings:")
print(loadings[["PC1", "PC2"]].round(3))
print(f"\nVariance captured by PC1 + PC2: {evr[0] + evr[1]:.1%}")
```

```python
pca2 = PCA(n_components=2, random_state=42)
P = pca2.fit_transform(X)

plt.figure(figsize=(8, 6))
plt.scatter(P[:, 0], P[:, 1], s=8, alpha=0.4, c="steelblue")
plt.xlabel("PC1"); plt.ylabel("PC2")
plt.title("All songs projected onto the first two principal components")
plt.show()
```

- Two principal components explain only roughly 41% of the variance; about seven components are needed to reach 94%, so the audio-feature space is genuinely high dimensional.
- PC1 captures overall intensity (strong loadings on energy, loudness and valence), while PC2 contrasts danceability against liveness and tempo.
- The PC1-PC2 scatter forms one continuous, overlapping cloud with no clearly separated blobs, an early sign that cluster boundaries will be soft rather than crisp.

## 5. K-Means clustering

The optimal number of clusters is chosen using the **elbow method** and the **silhouette score**.

```python
K_RANGE = range(2, 11)
wcss, kmeans_sils = [], []
for k in K_RANGE:
    km = KMeans(n_clusters=k, n_init=10, random_state=42)
    labels = km.fit_predict(X)
    wcss.append(km.inertia_)
    kmeans_sils.append(silhouette_score(X, labels))

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].plot(list(K_RANGE), wcss, "o-")
axes[0].set(title="Elbow method (WCSS)", xlabel="Number of clusters k", ylabel="WCSS")
axes[1].plot(list(K_RANGE), kmeans_sils, "o-", color="coral")
axes[1].set(title="Silhouette score vs k", xlabel="Number of clusters k", ylabel="Silhouette score")
plt.tight_layout()
plt.show()

for k, w, s in zip(K_RANGE, wcss, kmeans_sils):
    print(f"k={k}:  WCSS={w:,.0f}  |  silhouette={s:.3f}")
```

```python
K = 4
kmeans = KMeans(n_clusters=K, n_init=10, random_state=42)
km_labels = kmeans.fit_predict(X)
km_score = silhouette_score(X, km_labels)

print(f"K-Means silhouette (k={K}): {km_score:.3f}")
print("Cluster sizes:", pd.Series(km_labels).value_counts(sort=False).to_dict())
```

```python
km_profiles = data.copy()
km_profiles["Cluster"] = km_labels
print("Mean feature values per cluster:")
print(km_profiles.groupby("Cluster").mean().round(3).T)

plt.figure(figsize=(8, 6))
scatter = plt.scatter(P[:, 0], P[:, 1], c=km_labels, cmap="tab10", s=8, alpha=0.6)
plt.colorbar(scatter, ticks=range(K))
plt.xlabel("PC1"); plt.ylabel("PC2")
plt.title("K-Means clusters (k=4) in PCA space")
plt.show()
```

- WCSS drops sharply up to k = 4 and then flattens out, marking the elbow near 4 clusters.
- The silhouette score is highest at k = 2 (0.173) and falls to 0.152 at k = 4, so **k = 4 does not have the best silhouette**. All scores are low anyway, confirming weak separation in the data.
- k = 4 is therefore chosen on **utility, not on the metric**. Two clusters would only split the catalogue into a broad high-intensity / low-intensity pair, which is too coarse to seed playlists or recommendations. k = 4 sits at the elbow, keeps the silhouette within 0.02 of the peak, and separates four interpretable, actionable groups that the two-cluster solution hides:
  - **Cluster 0** - upbeat and danceable (high danceability and valence)
  - **Cluster 1** - spoken-word / rap-heavy (very high speechiness, most popular tracks)
  - **Cluster 2** - energetic and intense (highest energy and tempo, higher liveness)
  - **Cluster 3** - low-energy / mellow (lowest energy, loudness, tempo and valence)
  This is the standard metric-versus-purpose trade-off: the silhouette slightly prefers fewer clusters, but the value of the segmentation lies in distinct, nameable groups that correspond to real listening styles.

## 6. Hierarchical clustering

Ward's linkage is used, and the cluster hierarchy is inspected with a **dendrogram**.

```python
Z = linkage(X, method="ward")

plt.figure(figsize=(14, 6))
dendrogram(Z, truncate_mode="level", p=5, no_labels=True)
plt.title("Hierarchical clustering dendrogram (Ward linkage)")
plt.ylabel("Distance")
plt.show()
```

```python
H_RANGE = range(2, 8)
hier_sils = []
for k in H_RANGE:
    agg = AgglomerativeClustering(n_clusters=k, linkage="ward")
    hier_sils.append(silhouette_score(X, agg.fit_predict(X)))

for k, s in zip(H_RANGE, hier_sils):
    print(f"k={k}: silhouette={s:.3f}")

HK = 4
hier = AgglomerativeClustering(n_clusters=HK, linkage="ward")
hier_labels = hier.fit_predict(X)
hier_score = silhouette_score(X, hier_labels)
print(f"\nHierarchical silhouette (Ward, k={HK}): {hier_score:.3f}")
print("Cluster sizes:", pd.Series(hier_labels).value_counts(sort=False).to_dict())
```

```python
hier_profiles = data.copy()
hier_profiles["Cluster"] = hier_labels
print("Mean feature values per cluster:")
print(hier_profiles.groupby("Cluster").mean().round(3).T)

plt.figure(figsize=(8, 6))
scatter = plt.scatter(P[:, 0], P[:, 1], c=hier_labels, cmap="tab10", s=8, alpha=0.6)
plt.colorbar(scatter, ticks=range(HK))
plt.xlabel("PC1"); plt.ylabel("PC2")
plt.title("Hierarchical clusters (Ward, k=4) in PCA space")
plt.show()
```

- The dendrogram shows two broad families which subdivide into about four branches, supporting a k of 4.
- The silhouette score is **highest at k = 2 (0.189)** and only slightly lower at k = 4 (0.172), so k = 4 does **not** have the best silhouette. k = 4 is chosen on interpretability: it is the largest number of clusters that keeps the silhouette near its peak while yielding distinct, nameable segments.
- In particular, k = 4 isolates a "live recordings" group (liveness near 0.78) that K-Means merges into neighbouring clusters, plus a large mainstream cluster holding roughly 72% of songs - structure a two-cluster split would completely hide.

## 7. DBSCAN clustering

DBSCAN needs no pre-specified number of clusters, but its `eps` and `min_samples` must be tuned. A **k-distance plot** is used to guide the choice of `eps`.

```python
nn = NearestNeighbors(n_neighbors=15).fit(X)
distances, _ = nn.kneighbors(X)
kdist = np.sort(distances[:, -1])

plt.figure(figsize=(8, 5))
plt.plot(kdist)
plt.xlabel("Points sorted by distance")
plt.ylabel("15th nearest-neighbour distance")
plt.title("k-distance plot for choosing eps")
plt.show()

print("Parameter scan (eps, min_samples) -> clusters / noise / silhouette")
for eps in [1.1, 1.2, 1.3, 1.4]:
    for ms in [10, 15, 20]:
        db = DBSCAN(eps=eps, min_samples=ms).fit(X)
        n_cl = len(set(db.labels_)) - (1 if -1 in db.labels_ else 0)
        noise = (db.labels_ == -1).sum()
        sil = silhouette_score(X, db.labels_) if n_cl >= 2 else float("nan")
        print(f"eps={eps}, min_samples={ms}: clusters={n_cl}, noise={noise} ({noise/len(X):.0%}), silhouette={sil:.3f}")
```

```python
EPS, MIN_SAMPLES = 1.3, 15
dbscan = DBSCAN(eps=EPS, min_samples=MIN_SAMPLES)
db_labels = dbscan.fit_predict(X)

n_cl = len(set(db_labels)) - (1 if -1 in db_labels else 0)
noise = (db_labels == -1).sum()
db_score = silhouette_score(X, db_labels) if n_cl >= 2 else float("nan")

print(f"DBSCAN (eps={EPS}, min_samples={MIN_SAMPLES})")
print(f"Clusters found: {n_cl}  |  noise points: {noise} ({noise/len(X):.1%})")
print(f"Silhouette: {db_score:.3f}")
print("Cluster sizes:", pd.Series(db_labels).value_counts().to_dict())

plt.figure(figsize=(8, 6))
scatter = plt.scatter(P[:, 0], P[:, 1], c=db_labels, cmap="tab10", s=8, alpha=0.6)
plt.colorbar(scatter)
plt.xlabel("PC1"); plt.ylabel("PC2")
plt.title("DBSCAN clusters in PCA space (-1 = noise)")
plt.show()
```

- The k-distance curve rises smoothly with no clear knee, so there is no natural density threshold to separate points.
- With eps = 1.3 and min_samples = 15, DBSCAN finds only two clusters, flags about 20% of songs as noise, and still lands near the other models on silhouette (0.16).
- DBSCAN is the weakest fit here: the high-dimensional, fairly uniform data density does not suit density-based clustering.

## 8. Gaussian Mixture Models (GMM)

GMM fits a mixture of Gaussian distributions. **BIC / AIC** evaluate the number of components, with the silhouette score checked alongside.

```python
G_RANGE = range(1, 11)
bics, aics, gmm_sils = [], [], []
for g in G_RANGE:
    gm = GaussianMixture(n_components=g, random_state=42).fit(X)
    bics.append(gm.bic(X)); aics.append(gm.aic(X))
    gmm_sils.append(silhouette_score(X, gm.predict(X)) if g >= 2 else float("nan"))

fig, axes = plt.subplots(1, 2, figsize=(13, 5))
axes[0].plot(list(G_RANGE), bics, "o-", label="BIC")
axes[0].plot(list(G_RANGE), aics, "o-", label="AIC")
axes[0].set(title="BIC / AIC vs components", xlabel="Components", ylabel="Criterion score")
axes[0].legend()
axes[1].plot(list(G_RANGE)[1:], gmm_sils[1:], "o-", color="green")
axes[1].set(title="Silhouette vs components", xlabel="Components", ylabel="Silhouette")
plt.tight_layout()
plt.show()

for g, b, a, s in zip(G_RANGE, bics, aics, gmm_sils):
    extra = ", ".join(["silhouette = {:.3f}".format(s)]) if g >= 2 else ""
    print(f"g={g}:  BIC={b:,.0f}  |  AIC={a:,.0f}  |  {extra}")
```

```python
G = 2
gmm = GaussianMixture(n_components=G, random_state=42)
gmm_labels = gmm.fit_predict(X)
gmm_score = silhouette_score(X, gmm_labels)

print(f"GMM silhouette (g={G}): {gmm_score:.3f}")
print("Cluster sizes:", pd.Series(gmm_labels).value_counts(sort=False).to_dict())

gmm_profiles = data.copy()
gmm_profiles["Cluster"] = gmm_labels
print("Mean feature values per cluster:")
print(gmm_profiles.groupby("Cluster").mean().round(3).T)

plt.figure(figsize=(8, 6))
scatter = plt.scatter(P[:, 0], P[:, 1], c=gmm_labels, cmap="tab10", s=8, alpha=0.6)
plt.colorbar(scatter, ticks=range(G))
plt.xlabel("PC1"); plt.ylabel("PC2")
plt.title("GMM clusters (g=2) in PCA space")
plt.show()
```

- BIC and AIC decrease continuously as components are added and never bottom out, so neither criterion supports a small number of components - the data is not well modelled as a compact mixture of Gaussians.
- The silhouette score is highest at two components (0.10), which is the lowest of all six selected configurations, and it collapses as components increase.
- g = 2 is therefore selected on **parsimony and readability, not on the information criteria**. Because only two separable groups exist, extra components mostly fragment tiny, unstable margins instead of creating meaningful segments; the smallest readable model (g = 2) is the pragmatic choice. The BIC/AIC result is reported honestly as evidence that the dataset does not contain a clean Gaussian mixture.

## 9. Visualising clusters with t-SNE

t-SNE is a non-linear projection that tends to preserve local neighbourhood structure, giving a complementary view to the linear PCA projection.

```python
tsne = TSNE(n_components=2, init="pca", random_state=42)
T = tsne.fit_transform(X)

fig, axes = plt.subplots(1, 2, figsize=(15, 6))
for ax, labs, name in [
    (axes[0], hier_labels, "Hierarchical (Ward, k=4)"),
    (axes[1], km_labels, "K-Means (k=4)"),
]:
    scatter = ax.scatter(T[:, 0], T[:, 1], c=labs, cmap="tab10", s=6, alpha=0.6)
    ax.set_title(name)
    ax.set_xlabel("t-SNE 1"); ax.set_ylabel("t-SNE 2")
    plt.colorbar(scatter, ax=ax)
plt.tight_layout()
plt.show()
```

- Cluster regions overlap heavily in t-SNE space too, confirming that the audio features do not carve out well-separated groupings.
- Nearby songs cluster in small local neighbourhoods, but the overall layout is nearer to a continuum than to distinct islands.

## 10. Model comparison

```python
comparison = pd.DataFrame({
    "Model": ["K-Means (k=4)", "Hierarchical (Ward, k=4)", "DBSCAN (eps=1.3)", "GMM (g=2)"],
    "Clusters": [K, HK, n_cl, G],
    "Silhouette": [km_score, hier_score, db_score, gmm_score],
})
comparison["Silhouette"] = comparison["Silhouette"].round(3)
print(comparison.to_string(index=False))

plt.figure(figsize=(8, 5))
plt.bar(comparison["Model"], comparison["Silhouette"],
        color=["#4C72B0", "#55A868", "#DD8452", "#C44E52"])
plt.axhline(0, color="black", linewidth=0.8)
plt.ylabel("Silhouette score")
plt.title("Model comparison by silhouette score")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()
```

- Among the **six selected configurations**, Hierarchical (Ward, k=4) has the best silhouette (0.172), marginally ahead of K-Means (0.152).
- This table compares the chosen configurations, not every configuration tested. If all options were ranked by score alone, Hierarchical k = 2 (0.189) and K-Means k = 2 (0.173) would win - but those were rejected because a two-cluster split is too coarse to be useful. The comparison therefore reflects the best *practical* models, not the best raw scores.
- DBSCAN's 0.160 is misleading because it treats roughly 20% of the catalogue as noise; GMM trails with 0.101.
- All scores stay below 0.2, so every method produces overlapping, soft clusters rather than crisp groups - an honest limitation of clustering on audio features alone.

## 11. Final cluster interpretation

The hierarchical model (Ward, k=4) is used as the final solution. It does **not** have the best silhouette of every configuration tested (hierarchical k = 2 scores 0.189), but among the practical, multi-cluster options it scores highest (0.172) and, more importantly, it produces the most interpretable segments that separate into actionable groups. The choice balances the metric against segment utility rather than following the silhouette alone.

```python
sizes = hier_profiles["Cluster"].value_counts().sort_index()
for c in range(HK):
    print(f"Cluster {c}: {sizes[c]:,} songs ({sizes[c] / len(hier_profiles):.1%} of the catalogue)")

centroids = hier_profiles.groupby("Cluster")[AUDIO_FEATURES].mean()
centroids_z = (centroids - centroids.mean()) / centroids.std()

plt.figure(figsize=(11, 5))
sns.heatmap(centroids_z, cmap="coolwarm", center=0, annot=True, fmt=".2f",
            linewidths=0.5, cbar_kws={"label": "z-score vs. global mean"})
plt.title("Cluster profiles (z-scored feature means)")
plt.xlabel("Audio feature")
plt.ylabel("Cluster")
plt.tight_layout()
plt.show()
```

**Cluster 0 - Mainstream (72%):** near-average on every feature; an "average" sonic profile shared by the bulk of the catalogue.

**Cluster 1 - Low-energy / mellow (13%):** lowest energy, quietest (about -18 dB), slowest tempo and least positive mood. The label is deliberately cautious: the dataset has no `acousticness` column, so this group is described by what the data actually shows (low energy, quiet, slow) rather than labelled "acoustic" on assumption.

**Cluster 2 - Spoken-word / rap (9%):** highest speechiness (0.33) and danceability, and the most popular tracks; strongly supports vocal and rapped content.

**Cluster 3 - Live / concert recordings (6%):** liveness near 0.78 with high energy and a faster tempo; consistent with audience recordings, though `liveness` is a proxy measure rather than a direct "live audio" label.

### How the segments can be used in practice

- **Playlist generation:** each cluster is a ready-made seed list - "mellow study soundtrack" (Cluster 1), "focus-friendly spoken word" (Cluster 2), "energetic live sets" (Cluster 3) - that can be expanded by pulling more tracks from the same cluster without hand-curation.
- **Recommendations:** instead of matching on genre strings, a recommender can score a user's recent listens, find their dominant cluster, and surface nearby tracks *within that cluster* (e.g. other low-energy material for a user who mostly listens to calm music). This generalises to tracks the user has never seen.
- **Catalogue organisation:** the clusters give editorial staff coarse buckets for tagging and seasonal campaigns (e.g. promoting the low-energy cluster during study weeks) and expose underserved niches - Cluster 2's high popularity shows ready demand for spoken-word content that a coarse two-cluster model would never have surfaced.

## 12. Conclusion

- The Spotify audio features in this dataset form weak, heavily overlapping clusters; every model scored a silhouette below 0.2.
- Distance-based approaches (hierarchical and K-Means) work best, while DBSCAN and GMM underperform on this high-dimensional, near-uniform data.
- Model selection balanced metric and utility: the highest-scoring configurations (hierarchical and K-Means at k = 2) were deliberately rejected because a two-way high/low-intensity split is too coarse for real products, while k = 4 was kept because its four segments are nameable, stable and actionable.
- The four final segments - mainstream, low-energy/mellow, spoken-word/rap, and live/concert recordings - match intuitive musical styles, so they remain practical for playlist generation, recommendation and catalogue organisation even with modest separation.
